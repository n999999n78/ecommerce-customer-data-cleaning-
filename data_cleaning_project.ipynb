```python
import pandas as pd
data={'Customer_ID': [101, 102, 103, 103, 105, 106],
    'Name': ['Sara', 'Ahmed', 'Mona', 'Mona', None, 'Khaled'],
    'Total_Spend': [150.5, 200.0, None, 300.0, 450.0, 120.0],
    'City': ['Jeddah', 'Riyadh', 'Dammam', 'Dammam', 'Jeddah', None]}
df=pd.DataFrame(data)
print("---original Data Before Cleaning---")
print(df)
print("\n"+"="*40+"\n")
df_cleaned=df.drop_duplicates(subset=['Customer_ID'],keep='last').copy()
df_cleaned['Name']=df_cleaned['Name'].fillna('Unknown')
df_cleaned['City']=df_cleaned['City'].fillna('Not Specified')
mean_spend=df_cleaned['Total_Spend'].mean()
df_cleaned['Total_Spend']=df_cleaned['Total_Spend'].fillna(mean_spend)
print("---Cleaned Data After Processing---")
print(df_cleaned)
```
